## 3.3) Data cleaning - adding columns 

```python
df['Order Month'] = df['Order Date'].dt.to_period('M').astype('str')
```

```python
df['Order Year'] = df['Order Date'].dt.year
```

```python
df['Order Quarter'] = df['Order Date'].dt.to_period('Q').astype('str')
```

```python
df['Profit Margin'] = df['Profit'] / df['Sales'] 
```

## 4.1.1) KPI by Month

```python
# KPI calculation by month

KPI_monthly = df.groupby('Order Month').agg(
    Total_sales = ('Sales', 'sum'),
    Total_profit = ('Profit', 'sum'), 
    Unique_orders = ('Order ID', pd.Series.nunique) 
) 

KPI_monthly['Profit_margin'] = (KPI_monthly['Total_profit'] / KPI_monthly['Total_sales']) 
KPI_monthly['Average_order_value'] = (KPI_monthly['Total_sales'] / KPI_monthly['Unique_orders']).round(2)

KPI_monthly = KPI_monthly.reset_index()
KPI_monthly.head(5)
```

```python
df['Sales'].sum() 
```

## 4.1.2) Quarter over quarter growth 

```python
# Quarter over quarter 

KPI_quarter = df.groupby('Order Quarter').agg(
    Total_sales=('Sales', 'sum')
).sort_index()

# Calculate growth rate 

KPI_quarter['QoQ_Growth'] = KPI_quarter['Total_sales'].pct_change().round(4) 
KPI_quarter = KPI_quarter.reset_index()

KPI_quarter.head()
```

## 4.2.1) KPI by Category  

```python
# KPI calculation by Category 

KPI_category = df.groupby(['Category', 'Sub-Category']).agg(
    Total_sales=('Sales', 'sum'),
    Total_profit=('Profit', 'sum'),
    Total_quantity=('Quantity', 'sum')
)

KPI_category['Profit_margin'] = (KPI_category['Total_profit'] / KPI_category['Total_sales'])
    
KPI_category = KPI_category.reset_index()
KPI_category.head()
```

## 4.2.2) Top product 

```python
Top_products = df.groupby('Product Name').agg(
    Total_sales=('Sales', 'sum'),
    Total_profit=('Profit', 'sum'),
    Order_count=('Order ID', pd.Series.nunique)
).sort_values(by='Total_sales', ascending=False).head(10).reset_index() 
```

```python
Top_products.head(10)
```

## 4.3.1) KPI by Customer 

```python
# KPI calculation by customer 

KPI_customer = df.groupby('Customer ID').agg(
    Customer_name=('Customer Name', 'first'),
    Total_sales=('Sales', 'sum'),
    Total_profit=('Profit', 'sum'),
    Order_count=('Order ID', pd.Series.nunique),
    First_purchase=('Order Date', 'min'),
    Last_purchase=('Order Date', 'max')
)

KPI_customer['Avg_order_value'] = (KPI_customer['Total_sales'] / KPI_customer['Order_count']).round(2)
KPI_customer['Profit_margin'] = (KPI_customer['Total_profit'] / KPI_customer['Total_sales'])

KPI_customer = KPI_customer.reset_index()
KPI_customer.head()
```

## 4.3.1.1) Repeat customer ratio 

```python
## Repeat customer ratio 

# count of order by customer 
customer_order_count = df.groupby('Customer ID')['Order ID'].count()

# number of repeat customers 
repeat_customers = customer_order_count[customer_order_count > 1].count()

# total customers 
total_customers = customer_order_count.count() 

# ratio 
repeat_customer_ratio = round(repeat_customers / total_customers, 4)

print(f"Repeat Customer Ratio: {repeat_customer_ratio}") 
```

## 4.3.2) KPI by customer segment 

```python
# KPI calculation by Customer segment 

KPI_segment = df.groupby('Segment').agg(
    Total_sales=('Sales', 'sum'),
    Total_profit=('Profit', 'sum'),
    Order_count=('Order ID', pd.Series.nunique),
    Customer_count=('Customer ID', pd.Series.nunique) 
)

KPI_segment['Avg_order_value'] = (KPI_segment['Total_sales'] / KPI_segment['Order_count']).round(2)
KPI_segment['Profit_margin'] = (KPI_segment['Total_profit'] / KPI_segment['Total_sales'])

KPI_segment = KPI_segment.reset_index()
KPI_segment.head()
```

## 4.4) KPI by Region, State

```python
# KPI calculation by Region, State 

KPI_region = df.groupby(['Region', 'State']).agg(
    Total_sales=('Sales', 'sum'),
    Total_profit=('Profit', 'sum'),
    Order_count=('Order ID', pd.Series.nunique),
    Customer_count=('Customer ID', pd.Series.nunique) 
)

KPI_region['Avg_order_value'] = (KPI_region['Total_sales'] / KPI_region['Order_count']).round(2)
KPI_region['Profit_margin'] = (KPI_region['Total_profit'] / KPI_region['Total_sales'])

KPI_region = KPI_region.reset_index()
KPI_region.head()
```

## 5) Save to Excel 

```python
with pd.ExcelWriter("Monthly Sales Report.xlsx") as writer: 
    df.to_excel(writer, sheet_name="raw data", index=False)
    KPI_monthly.to_excel(writer, sheet_name = "Monthly", index=False)
    KPI_quarter.to_excel(writer, sheet_name = "Quarterly Growth", index=False) 
    KPI_category.to_excel(writer, sheet_name = "Category", index=False)
    Top_products.to_excel(writer, sheet_name = "Top Products", index=False) 
    KPI_customer.to_excel(writer, sheet_name = "Customer", index=False) 
    KPI_segment.to_excel(writer, sheet_name = "Segment", index=False) 
    KPI_region.to_excel(writer, sheet_name = "Region", index=False)
```
